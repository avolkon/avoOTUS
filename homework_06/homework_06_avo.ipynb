```python
def share_bread(N, K):
    """
    Делит куски эльфийского хлеба между хоббитами поровну.

    Args:
        N (int): Количество хоббитов
        K (int): Количество кусков хлеба

    Returns:
        tuple: (x, y) где:
            x - сколько кусков достанется каждому хоббиту
            y - сколько кусков останется в корзинке
    """
    # Целочисленное деление - сколько целых кусков каждому
    x = K // N

    # Остаток от деления - что останется в корзинке
    y = K % N

    # Возвращаем оба значения
    return x, y


# Проверка с примером из условия
result = share_bread(N=3, K=14)
print(f"Хоббитов: 3, кусков: 14")
print(f"Каждому: {result[0]}, в корзинке: {result[1]}")
print(f"Проверка: {result[0]} * 3 + {result[1]} = {result[0]*3 + result[1]}")

# Проверка через assert (не должно быть ошибок)
assert share_bread(N=3, K=14) == (4, 2)
print("Тест пройден!")

# Дополнительные тесты
print("\nДругие примеры:")
print(f"5 хоббитов, 17 кусков: {share_bread(5, 17)}")  # (3, 2)
print(f"2 хоббита, 5 кусков: {share_bread(2, 5)}")    # (2, 1)
print(f"4 хоббита, 8 кусков: {share_bread(4, 8)}")    # (2, 0)
```

```python
def leap_year(year):
    """
    Определяет, является ли год високосным (годом великого сражения).

    Високосный год:
    - делится на 4, но не делится на 100
    - или делится на 400

    Args:
        year (int): Год для проверки

    Returns:
        str: "YOU SHALL PASS" для високосного года
             "YOU SHALL NOT PASS" для невисокосного
    """
    # Проверяем условия високосного года
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return "YOU SHALL PASS"
    else:
        return "YOU SHALL NOT PASS"


# Проверка с примером из условия
assert leap_year(5) == 'YOU SHALL NOT PASS'
print("✓ Тест с годом 5 пройден")

# Дополнительные тесты
test_years = [5, 2020, 1900, 2000, 2024, 2023, 400, 100]

print("\nПРОВЕРКА ГОДОВ СРАЖЕНИЙ")
print("-" * 40)

for year in test_years:
    result = leap_year(year)
    print(f"Год {year}: {result}")

print("-" * 40)

# Подробный разбор для понимания
def explain_leap_year(year):
    """Показывает подробную проверку каждого года"""
    result = leap_year(year)
    print(f"\nГод {year}: {result}")

    if year % 400 == 0:
        print(f"  ✓ Кратен 400: {year} ÷ 400 = {year // 400} (остаток 0)")
    elif year % 100 == 0:
        print(f"  ✗ Кратен 100: {year} ÷ 100 = {year // 100} (остаток 0)")
        print(f"    Но не кратен 400 → НЕ високосный")
    elif year % 4 == 0:
        print(f"  ✓ Кратен 4: {year} ÷ 4 = {year // 4} (остаток 0)")
        print(f"  ✓ Не кратен 100: {year} ÷ 100 = {year // 100} (остаток {year % 100})")
    else:
        print(f"  ✗ Не кратен 4: {year} ÷ 4 = {year // 4} (остаток {year % 4})")

# Разбираем ключевые примеры
explain_leap_year(2020)  # Високосный (кратен 4, не кратен 100)
explain_leap_year(1900)  # Не високосный (кратен 100, но не 400)
explain_leap_year(2000)  # Високосный (кратен 400)
explain_leap_year(2023)  # Не високосный (не кратен 4)
explain_leap_year(5)     # Не високосный (из условия)
```

```python
def amulet_area(a, b, c):
    """
    Вычисляет площадь треугольника по трём сторонам (формула Герона).

    Аргументы:
        a (float): длина первой стороны
        b (float): длина второй стороны
        c (float): длина третьей стороны

    Возвращает:
        float: площадь треугольника
    """
    # Шаг 1: вычисляем полупериметр
    p = (a + b + c) / 2

    # Шаг 2: вычисляем подкоренное выражение
    under_root = p * (p - a) * (p - b) * (p - c)

    # Шаг 3: извлекаем квадратный корень (возводим в степень 0.5)
    S = under_root ** 0.5

    return S

# Проверка с примером из условия
assert amulet_area(3, 4, 5) == 6
print("✓ Тест с треугольником 3-4-5 пройден")

# Пошаговое объяснение для треугольника 3-4-5
def объяснить_вычисления(a, b, c):
    """Показывает все шаги вычисления площади"""
    print(f"\nПОДРОБНОЕ ВЫЧИСЛЕНИЕ ДЛЯ СТОРОН {a}, {b}, {c}")
    print("-" * 40)

    # Шаг 1
    p = (a + b + c) / 2
    print(f"1. Полупериметр p = ({a} + {b} + {c}) / 2 = {p}")

    # Шаг 2
    print(f"2. Вычисляем (p - каждую сторону):")
    print(f"   p - a = {p} - {a} = {p - a}")
    print(f"   p - b = {p} - {b} = {p - b}")
    print(f"   p - c = {p} - {c} = {p - c}")

    # Шаг 3
    under = p * (p - a) * (p - b) * (p - c)
    print(f"3. Подкоренное выражение:")
    print(f"   {p} * {p - a} * {p - b} * {p - c} = {under}")

    # Шаг 4
    S = under ** 0.5
    print(f"4. Квадратный корень: √{under} = {S}")
    print(f"5. Площадь = {S}")

# Показываем подробно для 3-4-5
объяснить_вычисления(3, 4, 5)

```

```python
import math
import random

def cal_euclidean(a, b):
    """
    Евклидово расстояние: √(Σ(aᵢ - bᵢ)²)
    """
    sum_squares = 0
    for i in range(len(a)):
        diff = a[i] - b[i]
        sum_squares += diff ** 2
    return math.sqrt(sum_squares)


def cal_manhattan(a, b):
    """
    Манхэттенское расстояние: Σ|aᵢ - bᵢ|
    """
    sum_abs = 0
    for i in range(len(a)):
        sum_abs += abs(a[i] - b[i])
    return sum_abs


def cal_cosine(a, b):
    """
    Косинусное расстояние: 1 - (a·b) / (||a|| * ||b||)
    """
    dot_product = 0
    norm_a_sq = 0
    norm_b_sq = 0

    for i in range(len(a)):
        dot_product += a[i] * b[i]
        norm_a_sq += a[i] ** 2
        norm_b_sq += b[i] ** 2

    norm_a = math.sqrt(norm_a_sq)
    norm_b = math.sqrt(norm_b_sq)

    if norm_a == 0 or norm_b == 0:
        cosine = 0
    else:
        cosine = dot_product / (norm_a * norm_b)

    return 1 - cosine


# Тестирование
print("-" * 60)
print("ВЫЧИСЛЕНИЕ РАССТОЯНИЙ МЕЖДУ ВЕКТОРАМИ")
print("-" * 60)

# Создаем случайные векторы
a = [random.randint(-10, 10) for _ in range(10)]
b = [random.randint(-10, 10) for _ in range(10)]

print(f"Вектор a: {a}")
print(f"Вектор b: {b}")
print("-" * 60)

# Вычисляем расстояния
euclidean = cal_euclidean(a, b)
manhattan = cal_manhattan(a, b)
cosine = cal_cosine(a, b)

print(f"Евклидово расстояние:     {euclidean:.4f}")
print(f"Манхэттенское расстояние: {manhattan}")
print(f"Косинусное расстояние:    {cosine:.4f}")
print("=" * 60)


# Пошаговое объяснение на маленьком примере
print("\n" + "=" * 60)
print("ПОДРОБНОЕ ОБЪЯСНЕНИЕ (на маленьких векторах)")
print("=" * 60)

# Берем маленькие векторы для наглядности
x = [3, 1, 2]
y = [1, 2, 1]

print(f"Вектор x = {x}")
print(f"Вектор y = {y}")
print("-" * 40)

# 1. Евклидово расстояние
print("1. ЕВКЛИДОВО РАССТОЯНИЕ:")
sum_sq = 0
for i in range(len(x)):
    diff = x[i] - y[i]
    sq = diff ** 2
    sum_sq += sq
    print(f"   x[{i}]-y[{i}] = {x[i]} - {y[i]} = {diff}, квадрат = {sq}")
print(f"   Сумма квадратов = {sum_sq}")
print(f"   Расстояние = √{sum_sq} = {math.sqrt(sum_sq):.4f}")
print()

# 2. Манхэттенское расстояние
print("2. МАНХЭТТЕНСКОЕ РАССТОЯНИЕ:")
sum_abs = 0
for i in range(len(x)):
    diff = x[i] - y[i]
    abs_diff = abs(diff)
    sum_abs += abs_diff
    print(f"   |x[{i}]-y[{i}]| = |{x[i]} - {y[i]}| = |{diff}| = {abs_diff}")
print(f"   Сумма модулей = {sum_abs}")
print()

# 3. Косинусное расстояние
print("3. КОСИНУСНОЕ РАССТОЯНИЕ:")
dot = 0
norm_x_sq = 0
norm_y_sq = 0

for i in range(len(x)):
    dot += x[i] * y[i]
    norm_x_sq += x[i] ** 2
    norm_y_sq += y[i] ** 2
    print(f"   Итерация {i}: x[{i}]*y[{i}] = {x[i]}*{y[i]} = {x[i]*y[i]}")

norm_x = math.sqrt(norm_x_sq)
norm_y = math.sqrt(norm_y_sq)
cosine = dot / (norm_x * norm_y)

print(f"\n   Скалярное произведение = {dot}")
print(f"   Норма x = √{norm_x_sq} = {norm_x:.4f}")
print(f"   Норма y = √{norm_y_sq} = {norm_y:.4f}")
print(f"   Косинус угла = {dot} / ({norm_x:.4f} * {norm_y:.4f}) = {cosine:.4f}")
print(f"   Косинусное расстояние = 1 - {cosine:.4f} = {1 - cosine:.4f}")
print("=" * 60)
```

```python
# import numpy as np

# ШАГ 1: Создаем матрицу 5×6 со случайными целыми числами от 0 до 50
my_array = np.random.randint(0, 51, size=(5, 6))

# ШАГ 2: Находим максимальный элемент во всей матрице
max_element = np.max(my_array)

# ШАГ 3: Находим индекс (номер столбца) максимального элемента
# np.argmax() ищет индекс максимального элемента в "плоском" представлении
# np.unravel_index() преобразует плоский индекс в индексы по размерностям (строка, столбец)
max_index = np.unravel_index(np.argmax(my_array), my_array.shape)
max_column = max_index[1]  # Берем номер столбца (индекс по второй оси)

# ШАГ 4: Выбираем всю колонку с максимальным элементом
selected_column = my_array[:, max_column]

# Выводим результаты
print('Shape: ', my_array.shape)
print('Array')
print(my_array)
print(f"\nМаксимальный элемент: {max_element}")
print(f"Находится в строке {max_index[0]}, столбце {max_index[1]}")
print(f"Колонка {max_index[1]}: {selected_column}")
```

```python
# Дополнительная проверка
print("\nПРОВЕРКА:")
print(f"Размер матрицы: {my_array.shape} (ожидаем 5×6)")
print(f"Все элементы в диапазоне [0,50]: {np.all((my_array >= 0) & (my_array <= 50))}")
print(f"Выбранная колонка имеет длину {len(selected_column)} (ожидаем 5)")
print(f"Максимальный элемент {max_element} находится в выбранной колонке: {max_element in selected_column}")
```
